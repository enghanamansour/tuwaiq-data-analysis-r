```r
# Q2: first six rows
head(data)

# Q3: column names
names(data)

# Q4: flower with longest Petal.Length
data[which.max(data$Petal.Length), ]

# Q5: average Sepal.Length per species
aggregate(Sepal.Length ~ Species, data = data, FUN = mean)

# Q6: plot Petal.Length vs Petal.Width colored by species
plot(
  data$Petal.Length, data$Petal.Width,
  col = data$Species,
  pch = 19,
  xlab = "Petal Length",
  ylab = "Petal Width",
  main = "Petal Length vs Petal Width by Species"
)
legend("topleft", legend = levels(data$Species), col = 1:3, pch = 19)
```
