```r
languages <- c("R", "Python", "Java", "C++", "JavaScript")
votes <- c(12, 18, 7, 6, 7)

# A different color for each language
bar_colors <- c("#0072B2", "#E69F00", "#009E73", "#CC79A7", "#D55E00")

# Task 1: Vertical bar chart
barplot(
  height = votes,
  names.arg = languages,
  col = bar_colors,
  main = "Favorite Programming Languages - Bar Chart",
  xlab = "Programming Language",
  ylab = "Number of Votes"
)

# Calculate each language's percentage of all votes
percentages <- votes / sum(votes) * 100
pie_labels <- sprintf("%s\n%.1f%%", languages, percentages)

# Task 2: Pie chart with language names and percentages
pie(
  x = votes,
  labels = pie_labels,
  col = bar_colors,
  main = "Favorite Programming Languages - Pie Chart"
)
```

```r
students <- data.frame(
  Name = c("Ali", "Sara", "Mona", "Omar", "Laila"),
  Age = c(20, 22, 21, 23, 20),
  Math = c(85, 90, 78, 88, 95),
  English = c(80, 85, 82, 75, 92)
)
print(students)

# Display the first 3 rows
head(students, 3)

# Display only the Name column
students$Name

# Add a new column called Total that contains Math + English
students$Total <- students$Math + students$English

# Add another column called Average
students$Average <- students$Total / 2

# Show only students whose average is greater than 85
subset(students, Average > 85)

# Find the highest Math grade
max(students$Math)

# Find the average age of the students
mean(students$Age)

# Sort the DataFrame by Average from highest to lowest
students <- students[order(-students$Average), ]
students

# Challenge: add a Result column - "Pass" if average >= 85, else "Needs Improvement"
students$Result <- ifelse(students$Average >= 85, "Pass", "Needs Improvement")
students
```
