```r
mean(dataset$HR_Salary, na.rm = TRUE)
median(dataset$HR_Salary, na.rm = TRUE)
min(dataset$HR_Salary, na.rm = TRUE)
max(dataset$HR_Salary, na.rm = TRUE)
sd(dataset$HR_Salary, na.rm = TRUE)
quantile(dataset$HR_Salary, 0.25, na.rm = TRUE)
quantile(dataset$HR_Salary, 0.75, na.rm = TRUE)
```

```r
hist(
  dataset$HR_Salary,
  main = "Distribution of Employee Salaries",
  xlab = "HR Salary",
  col = "#0072B2"
)
```

```r
avg_salary_dept <- aggregate(HR_Salary ~ Department, data = dataset, FUN = mean)

barplot(
  avg_salary_dept$HR_Salary,
  names.arg = avg_salary_dept$Department,
  main = "Average Salary by Department",
  xlab = "Department",
  ylab = "Average HR Salary",
  col = "#E69F00"
)
```

```r
boxplot(
  HR_Salary ~ Department,
  data = dataset,
  main = "Salary Distribution by Department",
  xlab = "Department",
  ylab = "HR Salary",
  col = "#009E73"
)
```

```r
plot(
  dataset$HR_Salary, dataset$Payroll_Salary,
  main = "HR Salary vs Payroll Salary",
  xlab = "HR Salary",
  ylab = "Payroll Salary",
  pch = 19,
  col = "#CC79A7"
)
```

```r
cor(dataset$HR_Salary, dataset$Payroll_Salary)
```

```r
hist(
  dataset$Compa_Ratio,
  main = "Distribution of Compa-Ratio",
  xlab = "Compa Ratio",
  col = "#D55E00"
)
```

From this analysis, HR_Salary and Payroll_Salary are almost identical and strongly positively correlated (0.9997), confirming payroll records match HR's official salary data. Salaries vary noticeably across departments and occupational levels, with Senior-level employees earning significantly more than Junior and Mid-level employees. The Markets department shows the highest average salary, while HR shows the lowest.
